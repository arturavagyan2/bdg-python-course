## Quick reference: `while`
A `while` loop checks its condition before each turn. Something inside must change the condition or use `break` to end the loop. Here the changing value is `minutes`.

````python
minutes = 2
while minutes > 0:
    print("Minutes:", minutes)
    minutes -= 1
````

## Quick reference: lists and `enumerate`
A list keeps several values in order, inside square brackets. A loop can visit those values without using their positions to look them up.

`enumerate()` gives you two values each turn: a number and the next item. `start=1` starts numbering at 1; without it, numbering starts at 0. Use two variable names before `in`. Run this example, then try a different starting number.

````python
books = ["Dune", "Coraline"]
for number, book in enumerate(books, start=1):
    print(number, book)
````

## Quick reference: `zip`
`zip()` pairs values by their order: first with first, second with second. Here each city is paired with its temperature. Use two variable names to receive the pair.

If one list is shorter, ordinary `zip()` stops when that list ends. It does not fill missing values or report an error. Keep related data in the same order.

````python
cities = ["Vanadzor", "Gyumri"]
temperatures = [8, 11]
for city, temperature in zip(cities, temperatures):
    print(city, temperature)
````
