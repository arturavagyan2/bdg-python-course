## W2 · Display text or calculate a number?
Suppose two workshop groups have two and three people. A written calculation helps someone see the counts; a calculated result tells them the total. Quoted text is copied literally. With unquoted numbers and +, Python performs the addition before print displays the result.

**Predict both lines.** Does print itself mean “calculate”?

```python
print("2 + 3")
print(2 + 3)
```

The first line shows `2 + 3`; the second shows `5`. The quotes change the meaning, even though the characters look similar. print displays the result it is given; it does not calculate words inside quotes. We are using only simple addition today; other operators and their ordering come in lesson 3.

## W4 · AI, examples and evidence
Artificial intelligence (AI) is the broad field of building systems for tasks such as recognizing speech, recommending items and generating language. Machine learning (ML) is an approach within AI: a system learns patterns from examples instead of having every decision explicitly written by a person.

Consider a spam filter. A written rule “flag messages containing free” makes a predictable decision about that word. A learned filter is trained on many messages labelled spam or legitimate and may use combinations of patterns. An example labelled “spam” tells the training process what kind of message it is; it is not another rule we type into this Python lesson. A model can still make mistakes, especially with examples unlike its training data. We cannot infer its exact answer just by saying it uses ML.

**Worked contrast:** a genuine message “Your parcel is ready” has no word free, so this rule does not flag it. A learned filter's answer would need to be tested. Now discuss **D1** with two new messages and suggest useful training evidence.

A large language model (LLM) learns patterns in text during training and uses them to generate responses. A fluent answer can sound like an explanation even when its claim is false. Fluency is a feature of the generated text, not a correctness check. It can help suggest ideas or explanations, but we must test claims we rely on.

**Synthetic demonstration:** an invented response says “`print(6 + 2)` gives 9.” First calculate it yourself, then run the exact code below. This is a deliberately false classroom example, not a model transcript.

```python
print(6 + 2)
```

Python prints 8. This fixed addition gives the same result on repeated runs; a confident generated sentence is not stronger evidence. We reject the specific claim because the exact code produces a different result. Testing one claim does not establish that every AI answer is wrong or right.
