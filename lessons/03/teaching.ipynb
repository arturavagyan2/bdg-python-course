## W1 · Store a value, then change it
A water station has eight bottles ready and two spare bottles. Names let us state those facts once and reuse them in a calculation.

`bottles_ready = 8` assigns the value on the right to the name on the left. `=` is an instruction to store a value, not a claim that two sides will always stay equal. Python evaluates the right side first. `total_bottles = bottles_ready + spare_bottles` calculates a number now and stores that result.

We can give print several arguments separated by commas. `print("Ready:", bottles_ready)` prints the label, a space and the current value. The variable name itself is not printed; putting it in quotes would print its letters.

### Names a reader can understand
Use letters, digits and underscores; a name cannot start with a digit. Names are case-sensitive: `bottles_ready` and `Bottles_ready` are different names. Spaces and hyphens do not belong in an ordinary variable name. Python also reserves words such as `class`; do not use reserved words as names. Choose lowercase words joined by underscores, called **snake_case**: `spare_bottles`. Do not replace useful built-ins such as print, input, type, int, float, str or bool with your own values. A readable name describes what its value means.

## W2 · Numbers and text are different values
A parcel code can be text even if every character is a digit. A weight needs a number if we will calculate with it. `type(value)` shows what kind of value Python currently has.

- **int:** whole numbers, such as a parcel count.
- **float:** numbers that can have a fractional part, such as a weight of 2.75.
- **str:** text, such as a parcel label or "20".
- **bool:** True or False, such as a stored yes/no fact. These two values have capital initials and no quotes. We do not make decisions with them until the next lesson.

**Predict:** which value below can be added numerically?

```python
parcel_count = 20
parcel_weight = 2.75
parcel_code = "20"
insured = True
print(type(parcel_count))
print(type(parcel_weight))
print(type(parcel_code))
print(type(insured))
print("Text joined:", "20" + "5")
print("Numbers added:", 20 + 5)
```

With two strings, + joins the text, called concatenation. With numbers it adds. Python does not decide that a parcel code ought to be a number. Choose a type based on what the value means and what you need to do with it.

### Convert deliberately
`int("20")` reads whole-number text as an integer. `float("2.75")` reads decimal text as a float. `str(20)` produces text for a label. Conversion returns a value; it does not silently rewrite a separate existing value. Assign the returned value when you need it later.

`int(2.9)` gives 2. Converting a float to int discards its fractional part toward zero; it does **not round** to the nearest whole number. For example int(-2.9) gives -2, not -3. Keep a measurement as float rather than accidentally throwing its fraction away.

```python
code_text = "20"
count_number = int(code_text)
weight_number = float("2.75")
label_text = str(count_number)
print(count_number, type(count_number))
print(code_text, type(code_text))
print(weight_number)
print(label_text, type(label_text))
print("Truncated:", int(2.9))
```

### A bool conversion checks emptiness or zero
For the simple values we use today: numeric zero becomes False, nonzero numbers become True; empty text becomes False and nonempty text becomes True. `bool("False")` is **True** because the string has characters. It does not interpret the meaning of the English word. Likewise, the text "0" is nonempty even though the number 0 is zero. Do not use bool(input(...)) as a yes/no-word interpreter.

**Predict each result, then run.** Afterward change only the quoted word to empty text and explain the change. This is a conversion demonstration, not an if/and/or lesson.

```python
print(bool(0))
print(bool(5))
print(bool(""))
print(bool("False"))
print(bool("0"))
```

## W4 · Choose the calculation that answers the question
An **expression** combines values, names and operators to produce a value. A label shop has 17 labels and each full pack holds 5. There are different useful questions:

- / divides, retaining a fractional result: 17 / 5 is 3.4.
- // gives the whole number of complete packs for these positive counts: 3.
- % gives what is left after those complete packs: 2 labels.

3 packs of 5 account for 15 labels; 2 remain. These answers mean different things. Do not use / and then silently truncate a measurement. In general // is floor division, rounding the quotient downward, including with float operands: 7 // 2.0 is 3.0. For negatives -7 // 2 is -4; int(-7 / 2) is -3 because truncation moves toward zero. We use positive counts in required tasks.

```python
labels = 17
labels_per_pack = 5
print("Division:", labels / labels_per_pack)
print("Full packs:", labels // labels_per_pack)
print("Labels left:", labels % labels_per_pack)
print("Labels in full packs:", (labels // labels_per_pack) * labels_per_pack)
```

### More than addition
`+` adds, `-` subtracts and `*` multiplies. Repeated identical quantities often need multiplication: 3 rows with 4 chairs per row have 3 * 4 chairs. `**` means a power, not multiplication by its right operand. A square display with 3 labels along each side has 3 ** 2 positions, which is 3 * 3. We use small powers with a clear meaning, not precedence puzzles.

**Predict:** distinguish 3 * 2 from 3 ** 2. Then trace the row/display example.

```python
rows = 3
chairs_per_row = 4
chairs = rows * chairs_per_row
chairs_used = 5
print("Chairs:", chairs)
print("Chairs remaining:", chairs - chairs_used)
display_side = 3
print("Square display positions:", display_side ** 2)
```

### A percentage is a share out of 100
A 5% fee means 5 for each 100. Divide the percentage by 100 to get the fraction, then multiply by the amount. For an amount of 200 and 5%, `200 * (5 / 100)` gives 10.0. This is a formula, not a special percent operator: Python's `%` means remainder, not percentage.

## W5 · Update a value and group the intended expression
When a value changes through arithmetic, `budget += 10` means `budget = budget + 10`. The old budget is read first; then a new value is stored under the same name. The same idea applies to -=, *= and /=. In the next example, a club receives 10, spends 4, gets matching funds that double the remainder and divides the resulting budget among 3 groups.

**Predict the value after each update.** /= performs ordinary division, so its result can be a float even if the earlier values were integers.

```python
budget = 30
budget += 10
print("After contribution:", budget)
budget -= 4
print("After spending:", budget)
budget *= 2
print("After matching funds:", budget)
budget /= 3
print("Per group:", budget)
```

### Parentheses state which values belong together
To average two reading sessions, add both times before dividing by 2. Python normally evaluates powers first, then multiplication/division/floor division/remainder, then addition/subtraction. Operators in the multiplication/division group and the addition/subtraction group normally work from left to right. Parentheses let us state a different grouping. For today's formulas, use parentheses whenever the intended group needs to be clear; we do not need a table of every Python operator.

**Predict both results:** the first line divides only the second session; the second divides their combined time. A program can run successfully while calculating a different question.

```python
first_minutes = 20
second_minutes = 40
print("Only second divided:", first_minutes + second_minutes / 2)
print("Mean session:", (first_minutes + second_minutes) / 2)
```
