## W1 · Ask a question about two values
A comparison produces True or False: a value we can store, print and later use to choose a path. At a workshop, 18 prepared kits are enough for 18 bookings. Equality matters at the boundary.

`==` asks “same value?”; `!=` asks “different?”; `>` and `<` mean strictly greater/smaller; `>=` and `<=` include equality. One `=` assigns a value to a name; two `==` compare values. `kits_ready = 18` stores a fact; `kits_ready == 18` asks about that fact.

**Predict all six:** which comparisons include the exactly-equal case?

```python
kits_ready = 18
bookings = 18
print("Equal:", kits_ready == bookings)
print("Different:", kits_ready != bookings)
print("More:", kits_ready > bookings)
print("Fewer:", kits_ready < bookings)
print("Enough:", kits_ready >= bookings)
print("At most:", kits_ready <= bookings)
```

### Strings compare as text
String equality checks characters, including their case: "ready" and "Ready" are different. String ordering compares from left to right, using the first differing character; if one is a complete prefix, the shorter comes first. For ordinary lower-case English examples this resembles dictionary order, but Python uses character values, not a language-aware dictionary.

In "10" versus "2", the first differing characters are "1" and "2". Python is not reading the text as the numbers ten and two. Convert numeric input before a numeric comparison.

**Predict:** compare the same written digits as text and as numbers. Then change the first text to "12" and check whether the text result changes.

```python
print("Same text:", "ready" == "Ready")
print("Different text:", "ready" != "Ready")
print("Word order:", "ant" < "bee")
print("Text order:", "10" < "2")
print("Number order:", 10 < 2)
print("Converted order:", int("10") < int("2"))
```

### Intentional error · order needs comparable types
The next cell deliberately raises TypeError: Python cannot order this integer and string as though they were two numeric quantities. This is different from equality: `10 == "10"` is simply False. Observe the error, then continue; do not add exception handling.

```python
print(10 < "2")
```

## W2 · Let a Boolean choose a block
A **block** is a group of instructions that belong together. `if condition:` checks the condition. If the comparison is True, Python runs the indented block; otherwise it skips that block. Use four spaces consistently. The colon starts the block; the unindented line after it is outside the decision.

A workshop must prepare extra kits only when ready kits are fewer than bookings. **Predict both printed lines with 12 kits and 14 bookings; then set bookings to 10. Which line still runs?**

```python
kits_ready = 12
bookings = 14
if kits_ready < bookings:
    print("Prepare extra kits")
print("Check complete")
```

### Choose between two paths with else
`else:` belongs to the if at the same indentation level. It runs when that if condition is false; else has no extra condition. For a valid numeric booking count, these paths cover both possibilities. The line after the decision runs either way.

```python
kits_ready = 12
bookings = 10
if kits_ready < bookings:
    print("Prepare extra kits")
else:
    print("Kits are sufficient")
print("Check complete")
```

### Intentional error · assignment is not a question
The next cell deliberately raises SyntaxError. A condition cannot use this assignment spelling. Repair it in a scratch copy with ==, then explain why the correct comparison is False for 12 and 10. This is an opportunity to read the final error line, not a new error-handling feature.

```python
kits_ready = 12
if kits_ready = 10:
    print("Exactly ten")
```

## W3 · Choose the first matching path
A fictional parcel service assigns exactly one route using whole-number weights from 1 to 10 kg:
- At least 5 kg: Large route.
- Otherwise at least 2 kg: Standard route.
- Otherwise: Small route.

`elif` means “otherwise, check this next condition”. An if/elif/else chain takes only the **first** matching branch; later tests are skipped. If no condition matches and there is no else, none of the branch bodies runs. With an else, one path is selected for these valid inputs.

**Predict:** 6 meets both numerical thresholds. Which label appears?

```python
weight_kg = 6
if weight_kg >= 5:
    print("Large route")
elif weight_kg >= 2:
    print("Standard route")
else:
    print("Small route")
```

### Two if statements ask two independent questions
The next program is intentionally runnable and prints two lines. Each if is a new decision. The else belongs only to the second if. This may be useful when several notices should appear, but it violates our “exactly one route” policy for 6 kg.

**Predict before running.** Do not assume an earlier true if suppresses another separate if.

```python
weight_kg = 6
if weight_kg >= 5:
    print("Large route")
if weight_kg >= 2:
    print("Standard route")
else:
    print("Small route")
```

### Priority and boundaries are part of the rule
If the >=2 test came first in the chain, 6 kg would take Standard and the Large branch would be unreachable: every value >=5 is also >=2. Fixing indentation alone would not fix the priority.

Manually predict **1, 2, 4, 5, 6** before changing weight_kg and running one case at a time. Expected labels are Small, Standard, Standard, Large, Large. These include below/at/above the thresholds; a case far from a boundary is not enough evidence. Practice **C** asks you to investigate a different set of ranges after the break.

## W4 · Combine requirements deliberately
A storage room opens only when a valid pass is present **and** the room is open. `and` requires both Boolean conditions. An information desk can be contacted by email **or** phone: `or` needs at least one, and both are also allowed. `not` reverses the truth value: not False is True; not True is False.

Use these operators on clear Boolean expressions. Mixed logic is easier to read when parentheses show which alternatives belong together. Python evaluates comparisons before not, then and before or, but write the intended groups instead of asking a beginner to decode a precedence puzzle.

```python
has_pass = True
room_open = False
print("Both required:", has_pass and room_open)
print("Either sufficient:", has_pass or room_open)
print("Room closed:", not room_open)
print("Neither fact changed:", room_open)
```

### Group alternatives before applying a restriction
A help desk serves someone with a booking OR a staff badge, provided access is NOT suspended. Suspension blocks both entry routes. The parentheses express that policy. Changing a Boolean with not in an expression does not reassign the original variable.

```python
has_booking = True
is_staff = False
is_suspended = True
can_enter = (has_booking or is_staff) and (not is_suspended)
print("Can enter:", can_enter)
```

### The repeated-comparison trap
`answer == "yes" or "y"` does **not** compare answer with two values. It combines the result of one comparison with the nonempty string "y". That string is truthy, so the if is taken even for "no". Write both comparisons: `(answer == "yes") or (answer == "y")`.

Truthiness is how a value behaves as a condition. For values taught so far, numeric zero and empty text are falsy; nonzero numbers and nonempty text are truthy. The string "False" has characters, so it is truthy. Use an explicit comparison when you mean the word's content. For this lesson we combine Boolean comparisons; in general and/or can return an operand rather than a bool, so do not treat them as automatic True/False converters.

```python
answer = "no"
if answer == "yes" or "y":
    print("Incorrect rule accepts no")
if (answer == "yes") or (answer == "y"):
    print("Correct rule accepts")
else:
    print("Correct rule rejects no")
```

```python
flag_text = "False"
if flag_text:
    print("Nonempty text enters the branch")
if flag_text == "True":
    print("The word is True")
else:
    print("The word is not True")
print("Empty text:", bool(""))
print("Zero:", bool(0))
```

## W5 · Nest when the second question depends on the first
A fictional equipment desk first checks training. Only trained people can borrow equipment; if trained, a reservation determines whether to collect the reserved kit or join the queue. Nesting keeps that second question inside the successful first path. Each inner block adds another four spaces. The inner else aligns with its inner if; the outer else aligns with the outer if.

```python
trained = True
has_reservation = False
if trained:
    if has_reservation:
        print("Collect reserved kit")
    else:
        print("Join equipment queue")
else:
    print("Training required")
```

### Simplify when only a combined yes/no matters
For a different question—“may this person collect a reserved kit now?”—the two requirements can be combined. These two versions are equivalent for that yes/no outcome. They are **not** equivalent to the earlier three-message desk program: the simplified versions deliberately do not explain which requirement is missing. Keep nesting when different explanations or follow-up actions matter.

```python
trained = True
has_reservation = True
if trained:
    if has_reservation:
        print("Collect now")
    else:
        print("Not ready")
else:
    print("Not ready")
if trained and has_reservation:
    print("Collect now")
else:
    print("Not ready")
```

## W6 · Write an explicit text rule, then question its meaning
For a fictional office inbox, use this exact priority:
1. If the message contains "urgent", label it Urgent review.
2. Otherwise if it contains "invoice", label it Accounts.
3. Otherwise label it Human review.

String membership, `"urgent" in message`, asks whether those characters appear consecutively anywhere in the string. It does not recognize whole words, meaning or negation. `message.lower()` returns a lowercase copy of the text; the dot connects the text to this built-in string operation, and the parentheses call it. Assign the returned copy to a name. It does not change the original text. We introduce only this one operation, not a full string-methods lesson.

**Predict:** the next entry contains both keywords. Which one has priority?

```python
message = input("Office message: ")
normalized = message.lower()
if "urgent" in normalized:
    print("Urgent review")
elif "invoice" in normalized:
    print("Accounts")
else:
    print("Human review")
```

### Rules and language-model proposals need different checks
An explicit rule is predictable: we can trace the exact matching characters and branch order. That does not guarantee its label matches the sender's intended request. A language model may recognize paraphrases and context but can also confidently misread negation or ambiguity. Check both against the message and the stated purpose; disagreement is evidence to inspect, not a vote on which tool is always right.
