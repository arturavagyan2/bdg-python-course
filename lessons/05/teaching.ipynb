## W2 · while checks again
while repeats while its condition is True. Change the counter inside the loop so the condition eventually becomes False.

```python
seconds = 3
while seconds > 0:
    print(seconds)
    seconds -= 1
print("Go!")
```

**Try:** Start at 5. If you remove seconds -= 1, this loop will not stop. Use Interrupt Kernel or Ctrl+C to stop a stuck run.

## W4 · keep asking
Use for when you have a sequence or a known count. Use while when stopping depends on what happens. while True keeps going until break runs.

```python
while True:
    answer = input("Type ready: ")
    if answer == "ready":
        print("Starting")
        break
    print("Try again")
```

**Try:** Enter wait, then ready. Keep the input inside the loop so each turn asks again. This is a practice command, not a login.

## W3 · continue in an input loop
You have used continue and break. Here they control repeated input: wait skips the message; done exits. Read fresh input at the top of every turn, so continue cannot reuse an old reply.

```python
while True:
    command = input("Command: ")
    if command == "wait":
        continue
    if command == "done":
        break
    print("Received:", command)
```

**Try:** Enter wait, hello, done. Which entries produce a message? If input moved above while, would a wait ever change?

## W1 · Reference: for
You have used for already. Use this example only if you need a reminder: the indented line repeats; Done prints once after the loop.

```python
for ticket in range(1, 4):
    print("Ticket", ticket)
print("Done")
```

## W1b · Reference: range and text
Quick reference, not a new topic: range leaves out its stop value. A for loop can also visit the characters of a string. Predict before running if you need a refresher.

```python
for number in range(2, 7, 2):
    print(number)
for letter in "BDG":
    print(letter)
```
