## W1 · a loop inside a loop
For each row, the inner loop starts again at seat 1. Two rows and three seats give six printed pairs.

```python
for row in range(1, 3):
    for seat in range(1, 4):
        print("Row", row, "Seat", seat)
```

## W3 · search and remember
found is a flag: a Boolean that remembers whether a match appeared. break stops at the first match. The final if handles the no-match case.

```python
found = False
for letter in "cat":
    if letter == "a":
        found = True
        print("Found a")
        break
if not found:
    print("No a")
```

**Try:** Try dog instead of cat. In nested loops, break stops only the loop it is inside.

## W4 · a menu repeats
Read the choice on every turn. The exit choice uses break; other choices return to the top. An unknown choice asks again.

```python
while True:
    choice = input("1: km to metres, 0: exit: ")
    if choice == "0":
        break
    elif choice == "1":
        km = float(input("Kilometres: "))
        print("Metres:", km * 1000)
    else:
        print("Choose 0 or 1")
```

## W4b · count words
text.split() separates text into words at spaces, tabs and newlines. Loop over those words and count them. Empty text gives no words. We will study lists in the next lesson.

```python
text = "Python is useful"
word_count = 0
for word in text.split():
    word_count += 1
print("Words:", word_count)
```

## W5 · plan the ATM
This is a fictional cash machine. Keep one balance outside the menu loop. Add a deposit; subtract a withdrawal only when there is enough money. No accounts, PINs or real money are involved.

```python
balance = 30
deposit = 10
balance += deposit
withdrawal = 15
if withdrawal <= balance:
    balance -= withdrawal
print("Balance:", balance)
```
