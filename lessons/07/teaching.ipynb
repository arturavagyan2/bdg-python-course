## W1 · Read part of a list
A list keeps several values in order. Index 0 is the first item; -1 is the last. A slice stops before its end index, so [1:3] selects indexes 1 and 2.

```python
items = ["tea", "rice", "soap", "apples"]
print(items[0], items[-1])
print(items[1:3])
print(items[:2])
print(items[2:])
items[1] = "pasta"
print(items)
print(len(items))
```

## W2 · Change a list
append adds at the end; insert chooses a position. remove deletes the first matching value; pop removes and returns an item (the last by default). sort changes the same list into ascending order. These methods change the list; append and sort return None, meaning no result value.

```python
queue = ["Ben", "Dina"]
print(queue)
queue.append("Eli")
print(queue)
queue.insert(1, "Cara")
print(queue)
queue.remove("Dina")
print(queue)
served = queue.pop(0)

print("Served:", served)
print("Waiting:", queue)
```

## W3 · Build a new list with a loop
Start with an empty result. Visit each price, calculate a new value and append it. The original prices stay unchanged.

```python
prices = [3, 6, 9]
reduced = []
for price in prices:
    reduced.append(price - 1)
print(reduced)
print(prices)
```

**Try:** Change the calculation to add 2. Explain which list changes.

## W3b · A short list comprehension
A comprehension builds a new list from a loop and an expression. Read [price - 1 for price in prices] as: for each price, keep price - 1. A final if keeps only matching items. Use a regular loop when several actions or input make the code easier to follow.

```python
prices = [3, 6, 9]
reduced = [price - 1 for price in prices]
small = [price for price in prices if price <= 6]
print(reduced)
print(small)
print(prices)
```

## W4 · Unpack a fixed pair
A tuple keeps ordered values whose positions cannot be reassigned. Use it for a fixed pair such as a room and its capacity; use a list for a changing queue. Unpacking puts each value into a separate name. The number of names must match the number of values. A one-item tuple needs a comma: ("Room A",).

```python
room = ("Room A", 12)
name, seats = room
print(name)
print("Seats:", seats)
print(room[0])
print(("Room B",))
```

**Try:** Change the capacity in the tuple literal and run again. Explain how replacing the whole tuple differs from changing one position.

## W5 · Keep a list between menu turns
This small playlist adds a title, displays the list or exits. The list is created before the loop, so earlier titles survive the next turn. Read the choice inside the loop. Your shopping manager will need more choices and a guarded removal. item in items checks whether a value is present in a list; check before remove so a missing value does not cause ValueError.

```python
titles = []
while True:
    choice = input("1 add, 2 show, 0 exit: ")
    if choice == "0":
        break
    elif choice == "1":
        title = input("Title: ")
        titles.append(title)
    elif choice == "2":
        print(titles)
    else:
        print("Unknown choice")
```
