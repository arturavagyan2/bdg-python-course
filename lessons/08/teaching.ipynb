## W1 · Keep each name once
A set keeps unique values. It is useful for attendance or tags when duplicates are unwanted. Sets have no defined display order or numeric indexes. sorted makes a separate list for predictable display; it does not change the set. Use set() for an empty set; {} is an empty dictionary.

```python
arrivals = ["Mia", "Noah", "Mia"]
present = set(arrivals)
print(sorted(present))
print(len(present))
print("Mia" in present)
present.add("Lena")
present.remove("Noah")
print(sorted(present))
```

## W2 · Compare two groups
Union keeps everyone in either group; intersection keeps only people in both; difference keeps people in the first group who are absent from the second. Difference has a direction. These methods return new sets and leave the original groups alone.

```python
morning = {"Ada", "Bo"}
evening = {"Bo", "Cy"}
print("Either:", sorted(morning.union(evening)))
print("Both:", sorted(morning.intersection(evening)))
print("Morning only:", sorted(morning.difference(evening)))
print("Evening only:", sorted(evening.difference(morning)))
```

## W3 · Look up a value by its key
A dictionary connects each unique key with a value. Here a product name is the key and its stock count is the value. Square brackets use the key, not a numeric position. Assignment adds a missing key or replaces its old value. del deletes one entry.

```python
stock = {"pens": 8, "pads": 3}
print(stock["pens"])
stock["pens"] = 6
stock["clips"] = 10
del stock["pads"]
print(stock)
print("pens" in stock)
print(6 in stock)
```

## W4 · Loop over dictionary entries
keys() provides keys, values() provides values and items() provides pairs. Unpack each pair into two names, as with a tuple. A normal for loop over a dictionary visits its keys. Dictionaries preserve insertion order, but lookup uses a key rather than its position. list(...) converts a keys/values view to a list for display.

```python
prices = {"tea": 2, "cake": 4}
print(list(prices.keys()))
print(list(prices.values()))
for product, price in prices.items():
    print(product, price)
```

## W5 · Check a key before lookup or deletion
Use membership before accessing a possibly missing key. This handles an expected missing entry without crashing. A fixed-key stock example differs from your contact-book task, which must read fresh names each turn.

```python
stock = {"pens": 8, "pads": 3}
product = "glue"
if product in stock:
    print(stock[product])
else:
    print("No such product")
if "pads" in stock:
    del stock["pads"]
print(stock)
```
