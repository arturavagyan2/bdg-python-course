## W1 · A list can hold rows
Each outer-list item is another list. Read one bracket at a time: seats[1] selects a row, then [0] selects a seat. Changing that seat keeps the other seats in place.

**Predict:** Which bracket chooses the row?

**Try:** Change the last seat in the first row. Predict which values remain unchanged.

## W2 · A dictionary gives each field a name
A list of dictionaries keeps several records in order. Each record has the same field names. tickets[1] selects one record; ["seats"] reads a field within it. A string key and a list index do different jobs.

**Predict:** What type of value is tickets[1]?

**Try:** Add a third fictional booking and display its fields using the same loop.

```python
tickets = [{"name": "Mia", "seats": 2},
           {"name": "Noah", "seats": 1}]
print(tickets[1]["name"])
tickets[1]["seats"] = 3
for ticket in tickets:
    print(ticket["name"], ticket["seats"])
```

## W3 · A dictionary can group lists
A dictionary value can be a list. A key selects a group and append changes that group’s list. A loop over items() gives the group name and its list, rather than one member at a time.

**Predict:** Does append add a new team key?

**Try:** Make the evening group empty and check the count.

```python
teams = {"morning": ["Mia"], "evening": ["Noah", "Bo"]}
teams["morning"].append("Lia")
for team, names in teams.items():
    print(team, len(names))
print(teams["morning"][1])
```

## W4 · Keys can lead through several levels
A value can itself be a dictionary. Follow the path from room to supplies to item. Each bracket works on the value produced by the previous bracket. Write a path on paper before editing a deep value.

**Predict:** Which keys exist directly inside rooms?

**Try:** Change open to False without replacing the whole Studio dictionary.

```python
rooms = {"Studio": {"supplies": {"pens": 4}, "open": True}}
print(rooms["Studio"]["supplies"]["pens"])
rooms["Studio"]["supplies"]["pens"] += 2
print(rooms["Studio"]["supplies"]["pens"])
```

## W5 · enumerate adds a position
enumerate gives a counter and the current item. By default the counter starts at zero and matches a list index. Starting at 1 is useful for display; that displayed number is then one more than the index.

**Predict:** What changes if the second argument is removed?

**Try:** Print the seat count beside each displayed booking number.

```python
tickets = [{"name": "Mia", "seats": 2},
           {"name": "Noah", "seats": 1}]
for number, ticket in enumerate(tickets, 1):
    print(number, ticket["name"])
```

## W6 · zip pairs matching positions
zip takes one item from each list per turn. It pairs positions, not matching names. It stops when the shorter list ends, so check lengths when every input must become a record. This lesson uses a length check, without exceptions.

**Predict:** Will Bo appear? Why?

**Try:** Add a seat count for Bo and run again.

```python
names = ["Mia", "Noah", "Bo"]
seats = [2, 1]
for name, count in zip(names, seats):
    print(name, count)
print("Lengths:", len(names), len(seats))
```

## W7 · Build a new shape for a different question
A list of records is useful for storing related fields. A list of names is useful for display. Build the names with a loop first, then compare the comprehension. Both create a new list; neither deletes records from stock.

**Predict:** Which part chooses the field, and which part filters the records?

**Try:** Change the zero to 2 and compare the two results.

```python
stock = [{"item": "pens", "count": 4},
         {"item": "pads", "count": 0}]
available = []
for row in stock:
    if row["count"] > 0:
        available.append(row["item"])
print(available)
print([row["item"] for row in stock if row["count"] > 0])
```

## W8 · Find a position before removing a record
A list has positions rather than name keys. Search the records and remember the matching index. After the search, remove that position only when a match exists. An index of -1 means not found in this program, so it must be checked before pop.

**Predict:** Why would pop(-1) be dangerous when a tag is missing?

**Try:** Search for an absent tag. The list must stay unchanged.

```python
items = [{"tag": "K1", "item": "keys"},
         {"tag": "B2", "item": "bag"}]
found_index = -1
for index, item in enumerate(items):
    if item["tag"] == "K1":
        found_index = index
        break
if found_index != -1:
    items.pop(found_index)
print(items)
```
