## W2 · Parameters receive values from a call
A parameter is a name in the definition. An argument is a value supplied by a call. Each call runs the same calculation with its own values. return sends the result back so the caller can save or combine it.

**Predict:** Which names are parameters? Which values are arguments?

**Try:** Change the second call to 5 tickets at a price of 2.

```python
def ticket_cost(count, unit_price):
    return count * unit_price

cost = ticket_cost(3, 4)
print(cost)
print(ticket_cost(2, 4) + 1)
```

## W3 · Printing does not return the printed number
print displays something for a person. A function with no return value gives its caller None, meaning no result was supplied. That is why displaying a calculation is not enough when later code needs the number.

**Predict:** Does result contain 12?

**Try:** Trace which print runs inside the function and which runs in the caller.

```python
def show_cost(count, price):
    print(count * price)

result = show_cost(3, 4)
print("Saved:", result)
```

## W3b · Return lets the caller choose what to do
return supplies a value without displaying it. The caller can print that value, add a fee, or compare it. Keep the calculation in one place so one change fixes every call.

**Predict:** Would a call on its own print anything in a script?

**Try:** Use the returned result in a comparison with 10.

```python
def calculate_cost(count, price):
    return count * price

result = calculate_cost(3, 4)
print("Saved:", result)
print("With fee:", result + 2)
```

## W4 · return ends the current call
A return immediately sends a value back to the caller. A later line in the function does not run on that path. Every call can choose a different path from its argument.

**Predict:** Does age 11 reach the second return?

**Try:** Check age 13 and explain the boundary.

```python
def entry_message(age):
    if age < 12:
        return "Come back later"
    return "Welcome"

print(entry_message(11))
print(entry_message(12))
```

## W5 · A loop finishes before its result returns
A function can contain a familiar loop. Put return after the loop when the result needs every item. Returning inside the loop would stop the whole function on its first iteration.

**Predict:** Why does an empty list return 0?

**Try:** Change the limit from 3 to 4 and test a four-letter word.

```python
def count_short_words(words):
    count = 0
    for word in words:
        if len(word) <= 3:
            count += 1
    return count

print(count_short_words(["map", "ticket", "bus"]))
print(count_short_words([]))
```

## W6 · A lookup function can return a record
Pass the records and search value into the function. Return the first match immediately; return None after the loop only when nothing matches. The caller uses is None to check for that missing result.

**Predict:** Why is return None after the loop?

**Try:** Try a missing name. is None checks for the special no-result value.

```python
def find_room(rooms, target):
    for room in rooms:
        if room["name"] == target:
            return room
    return None

rooms = [{"name": "Studio", "seats": 8}]
room = find_room(rooms, "Studio")
if room is None:
    print("Not found")
else:
    print(room["seats"])
```

## W7 · Each function has one job
A function name should say what it does. Here one function calculates a total and another displays a result. Passing values explicitly makes each part easier to check with small examples. DRY means Don’t Repeat Yourself: keep a repeated rule in one place.

**Predict:** Which function could a different display reuse?

**Try:** Try an empty booking list. Do not add input prompts inside total_seats.

```python
def total_seats(bookings):
    total = 0
    for booking in bookings:
        total += booking["seats"]
    return total

def show_total(total):
    print("Booked seats:", total)

bookings = [{"seats": 2}, {"seats": 3}]
show_total(total_seats(bookings))
```
