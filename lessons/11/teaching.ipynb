## W1 · Defaults and named arguments
A default supplies a value when a call omits that argument. A keyword argument names its parameter, so you can read the call without guessing the positions. Put positional arguments before keyword arguments.

**Predict:** Which calls use the default price? Does fee=0 use the default fee?

**Try:** Change only the second call to price=5. Explain which default still applies.


```python
def booking_cost(people, price=3, fee=1):
    return people * price + fee

print(booking_cost(2))
print(booking_cost(2, fee=0))
print(booking_cost(price=4, people=2, fee=1))
```

## W2 · Several positional values become a tuple
*durations collects extra positional arguments into a tuple. The name args is a convention, not a requirement. With no supplied values the tuple is empty, so an accumulator can return zero.

**Predict:** How many values enter the loop on the second call?

**Try:** Add a fourth duration. Keep the definition unchanged.


```python
def total_duration(*durations):
    total = 0
    for minutes in durations:
        total += minutes
    return total

print(total_duration(10, 5, 15))
print(total_duration())
```

## W3 · Named details become a dictionary
**details collects extra keyword arguments into a dictionary. Each key is the argument name. A record builder can keep useful details without requiring a fixed parameter for every possible field.

**Predict:** What type is details, and what happens when no details are supplied?

**Try:** Add language="English" at the call. Explain why False is still meaningful data.


```python
def make_badge(name, **details):
    return {"name": name, "details": details}

print(make_badge("Mina", role="host", paid=False))
print(make_badge("Bo"))
```

## W4 · A local name stays inside its call
Assigning a scalar parameter gives that call a new local value. The caller’s number stays the same until the caller saves a returned result. A name defined at module level is a global name.

**Predict:** Does the second print show 17 or 20?

**Try:** Save the returned value back into balance at the caller, then print it.


```python
balance = 20

def spend(balance, amount):
    balance = balance - amount
    return balance

print(spend(balance, 3))
print(balance)
```

## W5 · A global update affects later calls
global lets a function reassign a name from the module. This works, but the same call can depend on previous calls. Prefer passing the current value in and returning a new value when the function is a calculation.

**Predict:** Why does the same call return two different numbers?

**Try:** Compare this with next_visit(visits), which returns visits + 1 and leaves saving to the caller.


```python
visits = 0

def record_visit():
    global visits
    visits += 1
    return visits

print(record_visit())
print(record_visit())
```

## W6 · A function name can be a value
Pass the function name when another function should call it later. Adding parentheses calls it immediately. The receiving parameter can be used like any other function name.

**Predict:** Who calls double, and what value does that call receive?

**Try:** Define add_three(value) and pass its name to apply.


```python
def double(value):
    return value * 2

def apply(value, operation):
    return operation(value)

print(apply(4, double))
```

## W7 · A short expression can be a lambda
lambda makes a function from one expression. It returns that expression’s result. Use a named def when the rule needs explanation or several steps.

**Predict:** Which part is the parameter and which part computes the result?

**Try:** Replace the second lambda with a named double function. Compare readability.


```python
def apply(value, operation):
    return operation(value)

print(apply(4, lambda value: value + 3))
print(apply(4, lambda value: value * 2))
```
