## W5 · A date and a random classroom choice
date represents a calendar date. timedelta represents a span of time, so adding seven days also handles a month boundary. random.choice selects one supplied option. The seed below makes this demonstration repeatable; real classroom selection need not set a seed.

**Predict:** Will adding seven days keep the same month?

**Try:** Use a date near the end of December. Keep the choice list non-empty.


````python
from datetime import date, timedelta
import random

start = date(2026, 11, 28)
print(start + timedelta(days=7))
random.seed(7)
print(random.choice(["Mina", "Bo", "Nora"]))
````

## W6 · Paths describe locations without opening files
os.getcwd() gives the current working directory as text. Path turns a location into a path object. Its / operator joins path parts; name and suffix inspect the last part. Constructing a path does not create a file.

**Predict:** Does this create notes.txt?

**Try:** Build the path for a reports folder and summary.txt, then inspect its name without creating anything.


````python
import os
from pathlib import Path

folder = Path(os.getcwd())
print(folder == Path.cwd())
print((folder / "notes.txt").name)
print(Path("lesson.py").suffix)
````
